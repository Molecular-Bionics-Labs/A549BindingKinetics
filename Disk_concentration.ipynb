# Conversion from the micelle to disk 
1. micelle surface area: $$ S = 4\pi R^{2}$$
2. disk suface area: $$ S = 2\pi (R-t)^{2} + \pi t 2 \pi (R-t)$$

```python
import numpy as np 
R = 10
t_l = 5 
t = 6 

S_mic = 4 * np.pi * R**2 
S_disk = 2 * np.pi * (R-t)**2 + np.pi * t *2 * np.pi * (R-t)

c_mic= 2.35*10**-7
c_disk = c_mic*S_mic/S_disk
print(c_disk)
```
